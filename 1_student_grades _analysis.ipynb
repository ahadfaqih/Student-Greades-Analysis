# Student Greades Analysis

## Project Description
This project analyzes student grades using Python and basic data analysis techniques.

The dataset contains student scores in three subjects. The program calculates the average score for each subject, identifies the top-performing student, and visualizes the results using a bar chart.

## Goals:
- Practice working with tabular data.
- Calculate statistics such as averages.
- Visualize data with charts.

## Used Tools:
- Python
- Pandas
- Matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
data = pd.read_csv('/Students_Dataset.csv')
print(data)

# Calculate total grades
data['Total'] = data['Math Mark'] + data['Python Mark'] + data['English Mark']

# Find top student
top_student = data.loc[data['Total'].idxmax()]
print("\ntop_student:", top_student["Name"])

# Calculate averages
maths_avg = data['Math Mark'].mean()
python_avg = data['Python Mark'].mean()
english_avg = data['English Mark'].mean()

print("\nSubject Averages:")
print("Maths Average:", maths_avg ,"\nPython Average:", python_avg ,"\nEnglish Average:", english_avg)

# Visualization
plt.bar(['Math Mark', 'Python Mark', 'English Mark'], [maths_avg, python_avg, english_avg])
plt.xlabel('Subjects')
plt.ylabel('Average Score')
plt.title('Average Scores for Subjects')
plt.show()
```
